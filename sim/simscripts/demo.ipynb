In general, these are the main functions that memory.py uses to simulate a memory. Notice how only a few key parameters are utilized, those being:

- $\alpha$ : Also referred in code as SOF, like name suggests, adds onto decay rate, so higher SOF, the faster a memory is forgotten

- c : Introduced as an answer to spacing effect seen in memory. Memory introduced in a spaced manner tends to last longer, so c represents effective the spacing property is.

More parameter s and F while not used will be useful for [simulate.py](#simulate.py)
